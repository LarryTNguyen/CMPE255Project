# Data Cleaning

This notebook is primarily to explore and clean the datasets we are working on. 

## Citibike Data

```python
from pathlib import Path
import pandas as pd
```

```python


input_folder = Path(".\Datasets\Citibike")
output_file = Path("cleaned_combined.csv")

csv_files = list(input_folder.glob("*.csv"))

# delete old output if it exists
if output_file.exists():
    output_file.unlink()

first_chunk = True
dtype_map = {
    "start_station_id": "string",
    "end_station_id": "string"
}
for file in csv_files:
    print(f"Processing {file.name}...")

    for chunk in pd.read_csv(
        file,
        chunksize=100_000,   # adjust: 50k, 100k, 200k depending on memory
        dtype=dtype_map,
        low_memory=True
    ):
        # example cleaning
        chunk.columns = chunk.columns.str.strip().str.lower()

        # drop columns you do not need
        chunk = chunk.drop(columns=['start_station_id', 'end_station_id', 'ride_id', 'member_casual'], errors="ignore")

        # drop bad rows
        chunk = chunk.dropna()

        # example row filter
        # chunk = chunk[chunk["status"] == "available"]

        # write chunk directly to output
        chunk.to_csv(
            output_file,
            mode="w" if first_chunk else "a",
            header=first_chunk,
            index=False
        )

        first_chunk = False

print(f"Done. Cleaned file saved to: {output_file}")
```
